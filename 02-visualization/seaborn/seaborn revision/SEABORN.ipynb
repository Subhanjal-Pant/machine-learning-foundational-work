```python
flights = sns.load_dataset('flights')
```

```python
flights.head(2)
```

```python
flights['passengers'].value_counts().sum()
```

```python
flights['passengers'].sum()
```

```python
sns.catplot(x='month', y = 'passengers', kind = 'bar', data = flights)
```

```python
x=flights.pivot_table(index='year', columns='month', values='passengers', aggfunc='sum')
```

```python
sns.heatmap(x)
```

```python
sns.heatmap(x,cbar=False)
```

```python
sns.heatmap(x,cbar=False, linewidths=0.01)
```

```python
sns.heatmap(x,cbar=False, linewidths=0.01, annot= True, fmt='d')
```

```python
sns.heatmap(x,cbar=False, linewidths=0.01, annot= True, fmt='d', cmap='viridis')
```

```python
sns.heatmap(x,cbar=False, linewidths=0.01, annot= True, fmt='d', cmap='summer')
```

```python
sns.clustermap(x)
```

```python
sns.clustermap(x, metric="correlation", z_score=0)
```

```python
sns.clustermap(x, metric="correlation", z_score=0, annot=True)
```

```python
sns.clustermap(x, metric="correlation", z_score=0, annot=True, row_cluster=False)
```
