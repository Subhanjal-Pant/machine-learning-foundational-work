```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
tips.shape
```

```python
# scatter plot is an axis level function
sns.scatterplot(data = tips, x='total_bill', y = 'tip')
```

```python
#However, it can also be plotted as figure level function
#relplot comes in a squarerish shape
#The above graph and this graph are same however, width and height
# of figure are different
# figure level functions are recommended to be used
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter')
```

```python
sns.scatterplot(data = tips, x='total_bill', y = 'tip', hue='sex')
```

```python
sns.scatterplot(data = tips, x='total_bill', y = 'tip', hue='sex', style='time')
```

```python
sns.scatterplot(data = tips, x='total_bill', y = 'tip', hue='sex', style='time', size='size')
```

```python
#line plot
gap = px.data.gapminder()
gap.head()
```

```python
gap.shape
```

```python
temp_df = gap[gap['country']=='India']
temp_df.head()
```

```python
temp_df.shape
```

```python
#Axes level function
sns.lineplot(data=temp_df, x='year', y='lifeExp')
```

```python
#same graph using relplot
sns.relplot(data=temp_df, x='year', y = 'lifeExp', kind='line')
```
