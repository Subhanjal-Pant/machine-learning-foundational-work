```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("data.csv")
```

```python
data.head()
```

```python
data = data.drop(data.index[1])
data.head()
```

```python
data["Force"]=data["weight (kg)"] *9.81
```

```python
data.head()
```

```python
data['Displacement'] = 220 - data['distance (mm)']
```

```python
data.head()
```

```python
data["Stiffness(K)"] = (data['Force']/data['Displacement'])
```

```python
data.head()
```

```python
data
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
plt.plot(data["Displacement"], data["Force"])
plt.xlabel("Displacement")
plt.ylabel("Force")
plt.title("Force Vs Displacement")
```

```python
plt.plot( data["Displacement"],data["pressure (bar)"])
plt.xlabel("Displacement")
plt.ylabel("Pressure")
plt.title("Pressure Vs Displacement")
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
plt.plot( data["Force"],data["pressure (bar)"])
plt.xlabel("Force")
plt.ylabel("Pressure")
plt.title("Force Vs Pressure")
plt.grid()
plt.tight_layout()
plt.savefig("graph.pdf", format="pdf", bbox_inches='tight', pad_inches=0)
```
