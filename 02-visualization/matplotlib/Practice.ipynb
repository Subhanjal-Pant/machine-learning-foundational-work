```python
plt.plot(df['index'], df['RG Sharma'], color='green')
plt.plot(df['index'], df['V Kohli'], color='red')
plt.title("Rohit Sharma Vs Virat Kohli Comparision")
plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
plt.plot(df['index'], df['RG Sharma'], color='green', label="Rohit Sharma")
plt.plot(df['index'], df['V Kohli'], color='red', label='Virat Kohli')
plt.title("Rohit Sharma Vs Virat Kohli Comparision")
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.grid()
plt.legend()
```
