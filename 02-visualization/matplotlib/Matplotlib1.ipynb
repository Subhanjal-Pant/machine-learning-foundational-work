```python
plt.plot(batsman['index'], batsman['V Kohli'], color = 'green', linestyle = 'dashed', marker = 'D', label = 'Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color = 'black',linestyle = 'dashed', marker = '+', label = 'R Sharma')
plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()# yo vitra ni parameter pass garna milxa like upper right sty
```
