```python
def clean_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
```

```python
df['review'] = df['review'].apply(clean_html)
```

```python
df['review'].sample(1)
```

```python
def convert_lower(text):
    return text.lower()
```

```python
df['review'] = df['review'].apply(convert_lower)
```

```python
df['review'].sample(1).values
```

```python
def remove_special(text):
    x=''
    for i in text:
        if i.isalnum():
            x+=i
        else: 
            x=x+' '
      
    return x
```

```python
df['review'] = df['review'].apply(remove_special)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
len(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    x=[]
    for i in text.split():
        if i not in stopwords.words('english'):
            x.append(i)
        
    y = x[:]
    x.clear()
    return y
```

```python
df['review']=df['review'].apply(remove_stopwords)
```

```python
df
```

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()

y = []
def stem_words(text):
    for i in text:
        y.append(ps.stem(i))
    z = y[:]
    y.clear()
    return z
```

```python
df['review'] = df['review'].apply(stem_words)
```

```python
df.head(3)
```

```python
def join_back(list_input):
    return " ".join(list_input)
```

```python
df['review'] = df['review'].apply(join_back)
```

```python
df.head(2)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=1000)
```

```python
X = cv.fit_transform(df['review']).toarray()
```

```python
y = df.iloc[:, -1].values
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)
```

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
```

```python
clf1 = GaussianNB()
clf2 = MultinomialNB()
clf3 = BernoulliNB()
```

```python
clf1.fit(X_train, y_train)
clf2.fit(X_train, y_train)
clf3.fit(X_train, y_train)
```

```python
y_pred1 = clf1.predict(X_test)
y_pred2 = clf2.predict(X_test)
y_pred3 = clf3.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("Accuracy of GaussNB is :", accuracy_score(y_test,y_pred1)*100)
print("Accuracy of MultiNB is:", accuracy_score(y_test,y_pred2)*100)
print("Accuracy of BernauNB is :", accuracy_score(y_test,y_pred3)*100)
```
