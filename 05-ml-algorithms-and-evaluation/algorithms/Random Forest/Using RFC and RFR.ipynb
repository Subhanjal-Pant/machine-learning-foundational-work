```python
n_train = 150
n_test = 1000
noise =0.1
```

```python
def f(x):
    x=x.ravel()
    return np.exp(-x**2)+1.5*np.exp(-(x-2)**2)
```

```python
def generate(n_samples, noise):
    X = np.random.rand(n_samples)*10-5
    X=np.sort(X).ravel()
    y=np.exp(-X**2)+1.5*np.exp(-(X-2)**2)+np.random.normal(0.0,noise,n_samples)
    X=X.reshape((n_samples,1))
    return X,y
```

```python
X_train, y_train = generate(n_samples=n_train, noise=noise)
X_test,y_test = generate(n_samples=n_test, noise=noise)
```

```python
plt.figure(figsize=(18,8))
plt.plot(X_test, f(X_test), 'r')
plt.scatter(X_train, y_train, c='b', s=20)
plt.xlim([-5,5])
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dtree = DecisionTreeRegressor()
```

```python
dtree.fit(X_train,y_train)
```

```python
d_predict = dtree.predict(X_test)
```

```python
plt.figure(figsize=(18,8))
plt.plot(X_test, f(X_test), 'r')
plt.scatter(X_train,y_train, c='b', s=30)
plt.plot(X_test, d_predict, 'g', lw =4)
plt.xlim([-5,5])
plt.title("Decision Tree, MSE : %.2f" % np.sum((y_test-d_predict)**2))
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rfr = RandomForestRegressor(n_estimators=10000).fit(X_train,y_train)
```

```python
rf_predict = rfr.predict(X_test)
```

```python
plt.figure(figsize=(18,8))
plt.plot(X_test, f(X_test), 'r')
plt.scatter(X_train, y_train, c='b', s=30)
plt.plot(X_test, rf_predict, 'g', lw = 4)
plt.xlim([-5,5])
plt.title(f"Random Forest, MSE = {np.sum((y_test-rf_predict)**2)}")
```
