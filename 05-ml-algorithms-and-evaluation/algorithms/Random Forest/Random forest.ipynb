```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_features=5, n_informative=5, n_redundant=0,
                         n_clusters_per_class=1)
```

```python
df = pd.DataFrame(X, columns=['col1','col2', 'col3', 'col4','col5'])
df['target']=y
print(df.shape)
df.head(3)
```

```python
df.sample(5)
```

```python
def sample_rows(df, percent):
    return df.sample(int(percent*df.shape[0]), replace = True)
```

```python
import random
def sample_columns(df, percent):
    cols = random.sample(df.columns.tolist()[:-1], 
                         int(percent*df.shape[1]-1))
    new_df = df[cols]
    new_df['target']=df['target']
    return new_df
```

```python
df.columns
```

```python
df1 = sample_rows(df, 0.2)
df2 = sample_rows(df, 0.2)
df3 = sample_rows(df, 0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf1 = DecisionTreeClassifier()
clf2 = DecisionTreeClassifier()
clf3 = DecisionTreeClassifier()
```

```python
clf1.fit(df1.iloc[:,0:5], df1.iloc[:,-1])
```

```python
clf2.fit(df2.iloc[:,0:5], df1.iloc[:,-1])
```

```python
clf3.fit(df3.iloc[:,0:5], df1.iloc[:,-1])
```

```python
def combined_sampling(df, row_percent, col_percent):
    new_df = sample_rows(df, row_percent)
    return sample_columns(new_df, col_percent)
```

```python
combined_sampling(df, 0.5, 0.5).shape
```

```python
df1 = combined_sampling(df, 0.5, 0.5)
df2 = combined_sampling(df, 0.5, 0.5)
df3 = combined_sampling(df, 0.5, 0.5)
```
