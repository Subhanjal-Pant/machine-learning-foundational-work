```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn import datasets
```

```python
boston = pd.read_csv('BostonHousing.csv')
```

```python
boston.head(2)
```

```python
boston = boston.drop(columns='medv')
```

```python
boston.head(2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
```

```python
boston.iloc[:,0:-1]
```

```python
from sklearn.model_selection import train_test_split as tts
X_train,X_test,y_train,y_test=tts(boston.iloc[:,0:-1], boston.iloc[:,-1],
                                 test_size=0.2, random_state=123)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
knn = KNeighborsRegressor()
```

```python
lr.fit(X_train,y_train)
dt.fit(X_train,y_train)
knn.fit(X_train,y_train)
```

```python
y_pred1 = lr.predict(X_test)
y_pred2 = dt.predict(X_test)
y_pred3 = knn.predict(X_test)
```

```python
print("For Linear Regression: ", r2_score(y_pred1, y_test))
print("For Decision Trees: ", r2_score(y_pred2, y_test))
print("For KNN : ", r2_score(y_pred3, y_test))
```

```python
from sklearn.ensemble import BaggingRegressor
```

```python
bag_regressor = BaggingRegressor(random_state=1)
bag_regressor.fit(X_train,y_train)
```

```python
y_preds = bag_regressor.predict(X_test)
```

```python
print("Train Coeff : ", bag_regressor.score(X_train,y_train))
print("Test Coeff : ", bag_regressor.score(X_test,y_test))
```
