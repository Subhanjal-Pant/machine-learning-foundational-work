```python
class GDRegressor :
    def __init__(self, learning_rate, epochs):
        self.m = 88.54
        self.b = -120
        self.lr = learning_rate
        self.epochs = epochs
    def fit(self,X,y):
        for i in range(self.epochs):
            loss_slope = -2*np.sum(y - self.m*X.ravel() - self.b )
            self.b = self.b - (self.lr*loss_slope)
        print(self.b)
```

```python
gd = GDRegressor(0.001,1000)
```
