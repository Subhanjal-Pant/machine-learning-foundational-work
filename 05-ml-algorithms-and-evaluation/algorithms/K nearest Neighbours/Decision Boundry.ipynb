```python
import pandas as pd 
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
from sklearn.model_selection import train_test_split
```

```python
df.drop(columns = ['User ID', 'Gender'], inplace = True)
```

```python
X=df.iloc[:,0:2].values
```

```python
y=df.iloc[:,-1].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
a = np.arange(start=X_train[:,0].min()-1, stop = X_train[:,0].max()+1,
             step = 0.01)
```

```python
a
```

```python
a.shape
```

```python
b = np.arange(start=X_train[:,1].min()-1, stop=X_train[:,1].max()+1, 
             step=0.01)
```

```python
b.shape
```

```python
XX, YY = np.meshgrid(a,b)
```

```python
XX.shape
```

```python
YY.shape
```

```python
print(XX[0][0])
print(YY[0][0])
```

```python
K=17
```

```python
from sklearn.neighbors import KNeighborsClassifier
Knn = KNeighborsClassifier(n_neighbors=K)
```

```python
Knn.fit(X_train, y_train)
```

```python
Knn.predict(np.array([-2.938782990042754,-2.604255462878788]).reshape(1,2))
```

```python
input_array=np.array([XX.ravel(), YY.ravel()]).T
```

```python
y_pred = Knn.predict(input_array)
```
