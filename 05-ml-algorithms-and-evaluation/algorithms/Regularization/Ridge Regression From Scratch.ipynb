```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_regression
```

```python
X,y=make_regression(n_samples=100,n_features=1,n_informative=1,n_targets=1,
                   noise=20,random_state=13)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
l=LinearRegression()
```

```python
l.fit(X,y)
```

```python
l.coef_
```

```python
l.intercept_
```

```python
from sklearn.linear_model import Ridge
```

```python
rr = Ridge(alpha=10)

rr.fit(X,y)
```

```python
rr.coef_
```

```python
rr.intercept_
```

```python
rr1 = Ridge(alpha=100)

rr1.fit(X,y)
```

```python
rr1.coef_
```

```python
rr1.intercept_
```

```python
plt.scatter(X,y, color='blue')
plt.plot(X,l.predict(X), color='red', label='alpha=0')
plt.plot(X,rr.predict(X), color='green', label='alpha=10')
plt.plot(X,rr1.predict(X), color='yellow', label='alpha=100')

plt.legend()
```
