```python
from sklearn.model_selection import train_test_split as tts
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('train.csv', usecols=['Age','Fare','Survived'])
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
df.head(2)
```

```python
X=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
X
```

```python

y
```

```python
X_train,X_test,y_train,y_test=tts(X,y,test_size=0.2, random_state=42)
```

```python
X_train.head(2)
```

```python
clf = DecisionTreeClassifier()
```

```python
clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(clf,X_train,y_train,scoring='accuracy'))
```

```python
kbin_age=KBinsDiscretizer(n_bins=15, encode='ordinal', strategy='uniform')
kbin_fare=KBinsDiscretizer(n_bins=15,encode='ordinal',strategy='kmeans')
```

```python
trf=ColumnTransformer([
    ('first', kbin_age,[0]),
    ('second', kbin_fare,[1])
])
```

```python
X_train_trf=trf.fit_transform(X_train)
X_test_trf=trf.transform(X_test)
```

```python
trf.named_transformers_['first'].n_bins_
```

```python
trf.named_transformers_
```

```python
clf2 = DecisionTreeClassifier()
clf2.fit(X_train_trf,y_train)
```

```python
y_pred2=clf2.predict(X_test)
```

```python
accuracy_score(y_pred2, y_test)
```
