```python
# LOOCV-> Leave one out cross validation
#Supppose 10 rows of data is there
    # initially row will test data and remaining 9 will be train dat
    # then second row will be test and so on
# This method can be used when data is limited
```

```python
from sklearn.datasets import make_blobs
from sklearn.model_selection import LeaveOneOut
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
X,y = make_blobs(n_samples=10, random_state=1)
```

```python
cv=LeaveOneOut()
```

```python
X
```

```python
cv
```

```python
for i,j in cv.split(X):
    print(i,j)
```

```python
y_true,y_pred = [],[]
```

```python
for i,j in cv.split(X):
    print(i,j)
    X_train,X_test=X[i,:], X[j,:]
    print(X_train)
    print(X_test)
    y_train,y_test=y[i],y[j]
    model = RandomForestClassifier(random_state=1)
    model.fit(X_train,y_train)
    y_hat=model.predict(X_test)
    y_true.append(y_test[0])
    y_pred.append(y_hat[0])
```

```python
accuracy_score(y_true,y_pred)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
model1=RandomForestClassifier(random_state=1)
```

```python
scores=cross_val_score(model1,X,y,scoring='accuracy', cv=cv)
```
