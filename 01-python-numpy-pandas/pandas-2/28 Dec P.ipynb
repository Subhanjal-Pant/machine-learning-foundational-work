```python
import pandas as pd
import numpy as np
```

```python
country=['India', 'Pakistan', 'Usa', 'Nepal', 'Sri Lanka']
pd.Series(country)
```

```python
runs=[12,34,5,34,3,4]
pd.Series(runs)
```

```python
marks=[67,56,32,56,74,63]
sub=['English', 'Maths', 'Hindi', 'Science', 'Social', 'EPh']
pd.Series(marks,index=sub)
```

```python
marks=[12,34,24,52,55,56]
sub=['maths','EPH', 'Opt Maths', 'Social', 'Science','Nepali']
pd.Series(marks, index=sub, name='Subhanjal Marks')
```
